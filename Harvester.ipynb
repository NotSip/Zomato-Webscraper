# Importing Dependencies

```python
from selenium import webdriver #imported webdriver (main engine)
from selenium.webdriver.chrome.service import Service # Service helps in starting and stoping the chrome driver
from webdriver_manager.chrome import ChromeDriverManager # Automatically update chrome updates
from selenium.webdriver.common.by import By # Specifies how we want to find an element on the webpage
from selenium.webdriver.chrome.options import Options
import time
import pandas as pd
```

```python
options = Options()
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36")

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
```

# Intialising the Handshake

```python
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service = service)
```

# Visiting the URL


```python
url = "https://www.zomato.com/bhopal/restaurants"
driver.get(url)
```

```python
time.sleep(8)
```

# Harvesting Logic

```python
restaurant_links = set()
scroll_attempts = 30
```

```python
print("Starting the harvesting")
```

# Engine of harvest

```python
try:
    for i in range(scroll_attempts):
        # Scroll down by a specific amount 
        driver.execute_script("window.scrollBy(0, 1000);") # Scroll down by a specific amount 
        time.sleep(2) #Cooldown and gives site time to load the data
        
        anchors = driver.find_elements(By.TAG_NAME, 'a') # finds all the matching elements with tag a
        for a in anchors:
            link = a.get_attribute('href')
            # Look for restaurant order links in Bhopal
            if link and '/bhopal/' in link: # Helps in filtering the links contatining only bhopal and order in link
                restaurant_links.add(link)
        
        print(f"Progress: Scrolled : {i+1}/{scroll_attempts} | Unique Links found : {len(restaurant_links)}")
        df = pd.DataFrame(list(restaurant_links), columns=['restaurant_url'])
        df.to_csv('bhopal_raw_links.csv', index=False)
        print(f" Harvested {len(restaurant_links)} links ")
finally:
    driver.quit()
```
